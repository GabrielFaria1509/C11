Fundamentos de matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Plotando gráficos de linha

```python
##Eixos
x  = np.array([1,2,3,4])
y = x*2

y_2 = x**2


#Subplot(linha,coluna,id gráfico)
plt.subplot(1,2,1)
plt.plot(x,y,"*:r",linewidth = 3, markersize = 10)
plt.xlabel("x")
plt.ylabel("f(x)")

plt.subplot(1,2,2)
plt.plot(x,y_2,"s--r",linewidth = 3, markersize = 10)
plt.xlabel("x")
plt.ylabel("f(x)")

plt.show()
```

Plotando gráficos de dispersão

```python
ds = pd.read_csv("paises.csv",delimiter=";")

##Qual dispersa estão as rendas per capita dos 6 maiores países do mundo ?
##Pegando os 6 maiores do planeta

maiores = ds.nlargest(6,"Area (sq. mi.)")
print(f"Nome dos 6 maiores paises : {list(maiores["Country"])}")

#Gráfico
#s é uma terceira dimensão
plt.scatter(maiores["Country"],maiores["GDP ($ per capita)"],s=maiores["Area (sq. mi.)"]/10000)
```

Plotando gráficos de barras

```python
#Qual a diferença entre as 5 maiores rendas per capita
maiores_GDP = ds.nlargest(5,"GDP ($ per capita)")
print(f"Nome dos países com as maiores rendas per capita : {list(maiores_GDP['Country'])}")

#Gráfico
plt.bar(maiores_GDP["Country"],maiores_GDP["GDP ($ per capita)"])
```

Gráfico em torta

```python
condicao = ds['Coastline (coast/area ratio)'] == 0

paisesSemCosta = ds[condicao]

print(f"Países sem costa : {len(paisesSemCosta)}")

qtdComCosta = len(ds) - len(paisesSemCosta)

autopct = "%1.1f%%"
plt.pie(x=[qtdComCosta,len(paisesSemCosta)],labels=["Páises com costa","Países Sem costa"],autopct=autopct)

plt.show()
```
