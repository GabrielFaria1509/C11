Fundamentos de Seaborn

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ds_tips = sns.load_dataset('tips')
print(f"Importando dataset embutido no seaborn : ")
display(ds_tips)

sns.scatterplot(x='total_bill', y='tip', data=ds_tips)
plt.xlabel("Conta total em $US")
plt.ylabel("Gorjeta em $US")
plt.title("Conta x Gorjeta")
sns.set_style('dark')
sns.set_context("talk")
plt.show()
```

HISTOGRAMA

```python
ds_penguins = sns.load_dataset('penguins')
print(f"Importando dataset embutido no seaborn : ")
display(ds_penguins)

sns.histplot(data=ds_penguins, x='flipper_length_mm',hue="species",kde="True")
plt.show()
#
```

Boxplots

```python
ds_tips = sns.load_dataset('tips')

sns.boxplot(data = ds_tips, x='day', y='tip',hue="sex")
plt.show()
```

Regplot

```python
sns.regplot(data=ds_tips, x='total_bill', y='tip',line_kws={"color":"red"})
plt.show()
```

Heatmap

```python
corr = ds_tips[["total_bill","tip","size"]].corr()

sns.heatmap(corr,annot=True,fmt=".2f")

plt.show()
```
