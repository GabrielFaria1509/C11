```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('paises (2).csv',sep=';')

display(data)
```

```python
print(f"Colunas : \n{data.columns}")

print(f"Ver ponta superior : \n{data.head(5)}")

print(f"Ver ponta inferior : \n{data.tail(3)}")
```

Slicing no dataset

```python
print(f"Colunas : \n{data.columns}")

print(f"Pegando colunas específicas : \n{data[["Region","Climate","Service"]]}")
```

Separando arquivo original em novos arquivos

```python
total_population = np.sum(data["Population"])
print(f"Total população mundial : {total_population}")

seriesPercent = (data["Population"] / total_population) * 100

print("Adicionando porcentagem no dataset")
data["%Population"] = np.round(seriesPercent, 2)

display(data)

print("Criando nova versão do arquivo")
data.to_csv("paises_v2")

print(f"País com maior população : \n{data.nlargest(5,"%Population")}")
print(f"Nomes dos países com maior população : \n{data.nlargest(5,"%Population")["Country"]}")
print(f"Nome dos países com menor população : \n{data.nsmallest(5,"%Population")["Country"]}")
```

Agrupamento de dados - Dados categóricos

```python
print(f"Agrupando por região somando as medidas : \n{data.groupby("Region").sum()}")
grupo_region = data.groupby("Region")
print(f"Países por região : \n{grupo_region.count()["Country"]} ")
print(f"População por região : \n{grupo_region.sum()["Population"]}")
```

Criando funções customizadas

```python
def tenPercent(x):
  return x*0.9
```

```python
display(data.columns)

taxa_mortalidade = data["Deathrate"]

print(f"Aplicando função customizada : \n{taxa_mortalidade.apply(tenPercent)}")

print("Adicionando nova coluna no dataset")

data["Deathrate - 10%"] = taxa_mortalidade.apply(tenPercent)

display(data)
```
