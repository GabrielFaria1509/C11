-Trabalha normalmente com NumPy

-Series -> Armazena dados em uma dimensão, posso passar um dicionário pra alimentar ela, soma entre series ocorre por label

-Dataframe -> Armazena dados em duas dimensões, aceita dados heterogêneos, usado para datasets

```python
import pandas as pd
import math as mt
import numpy as np
```

```python
#Lista de label,lista de valores precisam ser passados pra series

labels = ["Faria","Turtle","Samuca","Zé"]
idade = [22,20,25,21]

se1 = pd.Series(index = labels, data = idade)
print(f"Tipo da variável : \n{type(se1)}")
print(f"Series completa : \n{se1}")

print(f"Valor guardado em Zé : \n{se1["Zé"]}")
print(f"Valor guardado em Faria e Turtle : \n{se1[["Faria","Turtle"]]}")
```

```python
series1 = pd.Series({"v1" : 10 , "v2" : 45, "v3" : 70})
series2 = pd.Series({"v1" : 10 , "v4" : 50, "v3" : 70})

print(f"Series 1 : \n{series1}")
print(f"Series 2 : \n{series2}")

print(f"Soma das series : \n{series1 + series2}")
print(f"Soma das series com uso de método  : \n{series1.add(series2,fill_value=0)}")



```

```python
#Lista de labels para linha e coluna
#Lista de valores

coluna = ["W","X","Y","Z"]
linhas = ["A","B","C","D","E"]

np.random.seed(10)
valores = np.random.randint(1,50,[5,4])
df = pd.DataFrame(columns=coluna,index=linhas,data=valores)

print(f"Dataframe gerado : \n{df}")

print(f"Gerando apenas uma coluna : \n{df["X"]}")
print(f"Gerando apenas duas colunas : \n{df[["Y","Z"]]}")
print(f"Gerando múltiplas colunas : \n{df[["W","X","Z"]]}")


print(f"Gerando apenas uma linha : \n{df.loc["A"]}")
print(f"Gerando apenas duas linhas : \n{df.loc[["B","E"]]}")
print(f"Gerando múltiplas linhas : \n{df.loc[["A","B","C"]]}")

#Passa coluna primeiro, depois linha
print(f"Gerando apenas uma célula \n{df["Y"]["C"]}")
print(f"Gerando apenas duas células \n{df[["Y","Z"]].loc[["A","B"]]}")
print(f"Gerando submatriz canto inferior direito : \n{df.loc[["D","E"],["Y","Z"]]}")

```

```python
coluna = ["W","X","Y","Z"]
linhas = ["A","B","C","D","E"]

np.random.seed(10)
valores = np.random.randint(1,50,[5,4])
df = pd.DataFrame(columns=coluna,index=linhas,data=valores)

print(f"Dataframe gerado : \n{df}")

print(f"Linha C : \n{df.iloc[2,:]}")

print(f"Linhas B,E : \n{df.iloc[[1,4],:]}")
print(f"Submatriz inferior canto direito : \n{df.iloc[[3,4],[2,3]]}")

```
