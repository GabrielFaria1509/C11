Aula 3 - Coleções

Tuplas

-Imutável, não coloco ou altero elementos

-Guarda coleção de elementos

-Permite slicing

```python
nomes = ("Goku","Trunks","Vegeta","Gohan")
print(nomes)
print(type(nomes))
#Comandos como nomes[4] = "Picollo" e nomes[0] = "Majin Boo"

#Slicing de dados(Técnica em análise de dados)
print(nomes[1])
print(nomes[1:3]) #O último número é exclusivo e o primeiro é inclusivo
print(nomes[1:])#Segundo em diante
print(nomes[-2])#Varre o vetor de trás pra frente
```

Listas

-Coleção mutável de elementos

-Permite slicing

```python
personagens = ["Dante","Vergil","Sparda","Nero"]
print(personagens)
print(type(personagens))
#CRUD de dados na lista
#CREATE
personagens.append("Lady") #adiciona no final
personagens.insert(2,"Triss")#Seleciona onde quero adicionar
print("Lista após inserir Triss e Lady : {} ".format(personagens))
#READ(Mesmo procedimento da tupla)
print("Primeiro da lista após inserção : {} ".format(personagens[1]))
print("Primeiro até o segundo : {} ".format(personagens[1:3])) #O último número é exclusivo e o primeiro é inclusivo
print("Segundoe em diante : {} ".format(personagens[1:]))#Segundo em diante
print("Varrendo de trás pra frente : {}".format(personagens[-2]))
#UPDATE
personagens[0] = "Yamato"
print("Lista após atualizar com Yamato : {} ".format(personagens))
#DELETE
personagens.remove("Yamato")
print("Conjunto após remover a yamato : {} ".format(personagens))


```

Conjuntos

-Não guarda ordem dos elementos

-Não guarda elementos duplicados

-permite naturalmente realizar operações entre conjuntos de dados(interseção,união,diferença)


```python
protagonistas = {"Naruto","Asta","ichigo"}
print(type(protagonistas))
print(protagonistas)
#CRUD

#CREATE
protagonistas.add("Goku") #add é apenas de conjunto
print("Conjunto após adicionar o Goku : {} ".format(protagonistas))
```

Dicionários

-Organiza os elementos

-Coleção mutável

```python
individuo_goku = {
    "nome" : "Goku",
    "Idade" : 52,
    "sexo" : "M"
}
#READ
print(individuo_goku)

#CREATE
individuo_goku["Obra"] = "Dragon Ball"
print("Dicionário após adicionar a obra : {} ".format(individuo_goku))

#UPDATE
individuo_goku["Obra"] = "Dragon ball GT"
print("Dicionário após atualizar a obra : {} ".format(individuo_goku))

#DELETE
del individuo_goku["sexo"]
print("Dicionário após remover o sexo : {} ".format(individuo_goku))
```

Posso misturar as diferentes coleções, como lista de dicionários


```python
pessoa = {
    "Nome " : "Faria ",
    "cidade" : "Três Corações"
    }
pessoa2 = {
    "Nome " : "Bebella ",
    "cidade" : "Três Corações"
    }
pessoa3 = {"Nome " : "Miria ",
           "cidade" : "Aiuruoca"
           }

serumaninhos = [pessoa,pessoa2,pessoa3] # Lista de dicionários

print("Lista de dicionários completa  {}: ".format(serumaninhos))

print("Imprimindo um elemento : {}".format(serumaninhos[1]))

print("Imprimindo apenas uma chave de um elemento : ".format(serumaninhos[1]["cidade"]))

print("Imrpimindo segundo dicionário pra frente : {}".format(serumaninhos[1:]))


```
