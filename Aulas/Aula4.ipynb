Aula 4 - Numpy

-Usa estrutura própria ndArray(Numpy Array), usa normalmente tuplas e listas

-Numpy é muito otimizado(usa C no backend)

```python
import numpy as np
#Importando biblioteca numpy
```

```python
arr = np.array([10,20,30,40,50,60])
print(type(arr))
print(arr)
print(f"tamanho array : {arr.size}")
print(f"Dimensões do array(1,2,3) : {arr.ndim}")
print(f"Formato do array(3 x 2, 1 x 0) : {arr.shape} ")

mtz = np.array([[10,20],
                [30,40],
                [50,60]])
print(type(mtz))
print(f"Matriz : \n{mtz}")
print(f"tamanho matriz : {mtz.size}")
print(f"Dimensões da matriz : {mtz.ndim}")
print(f"Formato da matriz : {mtz.shape}")
```

Funções pré-prontas

```python
mtz = np.ones([5,5])
print(f"Matriz de um :  \n{mtz}")

mtz_zeros = np.zeros([5,5])
print(f"Matriz de zeros : \n{mtz_zeros}")

mtz_eye = np.eye(5)
print(f"Matriz identidade : \n{mtz_eye}")

arr = np.zeros([10])
print(arr.reshape([5,2]))

print(f"Array : {arr}")
print(f"Array transformado em matriz : {arr}")

mtz = np.arange(2,21,2)
print(f"Array de 2 a 20 de 2 em 2 : {mtz}")

print(f"Array transformado em matriz :  \n{mtz.reshape(2,5)}")


```

Operações entre Numpy arrays

```python
arr1 = np.array([10,20,30,40,50])
arr2 = np.array([1,2,3,4,5])

print(f"Soma : {arr1 + arr2}") # ou arr3 = arr1 + arr2
print(f"Subtração : {arr1 - arr2}")
print(f"Multiplicação : {arr1 * arr2}")
print(f"Divisão : {arr1 / arr2}")
print(f"Potência : {arr1 ** arr2}")
print(f"Raiz quadrada : {np.sqrt(arr1)}")
print(f"Exponencial : {np.exp(arr1)}")
print(f"Logaritmo : {np.log(arr1)}")

arr3 = np.concatenate([arr1,arr2])
print(f"Arrays concatenados : {arr3}")

print(f"Escalar vezes o array : {2 * arr1}") # Broadcasting





```

```python
mtz = np.arange(10,96,5)
print(f"Array gerado : {mtz}")
print(f"Array transformada em matriz  : \n{mtz.reshape(3,6)}")
mtz = mtz.reshape(3,6)

#Extraindo soma das colunas
print(f"Soma colunas  : {mtz.sum(axis=0)}")

print(f"Soma linhas  : {mtz.sum(axis=1)}")

print(f"Soma primeira coluna : {mtz.sum(axis=0)[0]}")

print(f"Soma segunda linha : {mtz.sum(axis=1)[1]}")

print(f"Média das colunas : {mtz.mean(axis=0)}")

print(f"Média das linhas : {mtz.mean(axis=1)}")

print(f"Pegando soma de segunda coluna em diante : {mtz.sum(axis = 0)[1:,]}")

print(f"Pegando soma de segunda linha em diante : {mtz.sum(axis = 1)[1:,]}")





```

Núemros aleatórios no Numpy

```python
np.random.seed(5)
#Coloca uma semente que garante que os números serão os mesmos para todas máquinas
arr = np.random.randint(10)
print(f"Gerando número aleatório entre 0 e 10 : {arr}")

arr = np.random.randint(1,10,10)

print(f"Delimitando o sorteio(1 a 10,10 números) : {arr}")

print(f"Transformando em um conjunto : {set(arr)}")

print(f"Transformando em um conjunto(usando numpy) : {np.unique(arr)} ")

elementos_unicos, contagem = np.unique(arr,return_counts=True)

print(f"Elementos únicos : {elementos_unicos}")
print(f"Contagem : {contagem}")

```
