```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
dataset_air_travel = pd.read_csv("datasets/airtravel.csv",parse_dates=["Date"],index_col="Date")["Passengers"]

plt.figure(figsize=(12,6))
plt.plot(dataset_air_travel, marker='o', linestyle='-', color='blue')
plt.title("Passageiros de Voo Internacional Mensal (1949-1960)")
plt.xlabel("Data");plt.ylabel("Número de Passageiros");plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
dec_air = seasonal_decompose(dataset_air_travel, model='multiplicative', period=12)
dec_air.plot()
plt.suptitle("Decomposição da Série Temporal de Passageiros de Voo Internacional", fontsize=16)
plt.tight_layout()
plt.show()
```

A tendência é crescente
SAZONALIDADE: Sim. Período de 12 meses (anual). O pico ocorre em
julho/agosto (férias de verão) e o vale em novembro. O modelo é
multiplicativo, pois a amplitude das oscilações aumenta com o nível da série.
CICLO: Não. Com apenas 36 observações (3 anos) não há dados suficientes
para identificar oscilações de longo prazo sem período fixo. Todo o
padrão repetitivo é explicado pela sazonalidade anual.


```python
dataset_emissions = pd.read_csv("datasets/co2_emissions.csv",index_col="Year")["CO2_Emissions"]
dataset_emissions_index = pd.to_datetime(dataset_emissions.index, format="%Y")

plt.figure(figsize=(12,6))
plt.plot(dataset_emissions_index, dataset_emissions, marker='o', linestyle='-', color='green')
plt.title("Emissões de CO2 (1960-2020)")
plt.xlabel("Ano");plt.ylabel("Emissões de CO2 (milhões de toneladas)");plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
dec_co2 = seasonal_decompose(dataset_emissions, model='additive', period=8)
dec_co2.plot()
plt.suptitle("Decomposição da Série Temporal de Emissões de CO2", y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

Tendência decrescente
SAZONALIDADE: Não. Os dados são anuais (1 ponto por ano), então não existe
padrão que se repita dentro do ano. Sazonalidade exigiria dados mensais/trimestrais.
CICLO: Sim. Há oscilações em torno da tendência, com picos em ~1992, ~2000,
2007 e 2015 (período aproximado de 7-8 anos)
