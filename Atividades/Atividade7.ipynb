```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ds = pd.read_csv("paises.csv",delimiter=";")
print(f"Colunas do dataset : \n{ds.columns}")

display(ds)
```

Questão 1

Por meio do dataset paises.csv, trace dois gráficos de linhas em
um mesmo plano cartesiano, um mostrando a taxa de mortalidade
(Deathrate) e outro a taxa de natalidade (Birthrate) dos países da
América do Norte


```python
#Pegando região(Retirando espaços em branco)
ds['Region'] = ds['Region'].str.strip() #Método do panda mesmo pra tirar espaço em branco
north_america_countries = ds[ds['Region'] == 'LATIN AMER. & CARIB']
#Gerando primeiro gráfico(Deathrate e Birthrate)
plt.figure(figsize=(10,6))
plt.plot(north_america_countries['Country'],north_america_countries['Deathrate'],label="Deathrate")
plt.plot(north_america_countries['Country'],north_america_countries['Birthrate'],label="Birthrate")
plt.xlabel("Country")
plt.ylabel("Rate")

plt.title("Deathrate e Birthrate dos países da América do Norte")
plt.legend()
plt.show()




```

```python
data = pd.read_csv("space.csv",delimiter=";")
print(f"Colunas do dataset : \n{data.columns}")

display(data)
```

Questão 2

Por meio do dataset space.csv, trace um gráfico em barras
mostrando quantas empresas espaciais diferentes os EUA e a CHINA

Dica: não se esqueça de retirar os resultados repetidos


```python
#Deixando dataset só com China e Estados Unidos

mask_USA= data['Location'].str.contains("USA")
mask_China = data['Location'].str.contains("China")

dados_EUA = data[mask_USA]
dados_China = data[mask_China]

empresas_EUA = dados_EUA['Company Name'].nunique()
empresas_China = dados_China['Company Name'].nunique()

plt.bar(["EUA","China"],[empresas_EUA,empresas_China],color = ["blue","red"])
plt.title("Empresas únicas de cada país")

plt.show()
```

Questão 3

Por meio do dataset space.csv, trace um gráfico em torta
ilustrando a porcentagem de missões da empresa Roscosmos que
deram certo e que deram errado;



```python
mask_empresa = data['Company Name'] == "Roscosmos"

dados_empresa = data[mask_empresa]

acertos = dados_empresa['Status Mission'] == "Success"
erros = dados_empresa['Status Mission'] == "Failure"

n_acertos = acertos.sum()
n_erros = erros.sum()

porcentagem_acertos = (n_acertos / (n_acertos + n_erros)) * 100
porcentagem_erros = (n_erros / (n_acertos + n_erros)) * 100

#Deixei label de fora do .pie para legenda ficar somente na caixa

fatias = plt.pie([porcentagem_acertos,porcentagem_erros],autopct="%1.1f%%")
plt.legend(labels=['Sucessos', 'Falhas'])
plt.title('Porcentagem de acertos e erros da empresa Roscosmos')
plt.show()




```

Questão 4

Utilizando o dataset space.csv, trace um gráfico em barras com as
5 empresas que mais tiveram missões com status “Failure”;


```python
mask_failure = data['Status Mission'] == "Failure"

dados_failure = data[mask_failure]

mais_falhas = dados_failure["Company Name"].value_counts().nlargest(5)
#.value_counts() : você tem uma coluna com várias categorias misturadas e quer saber a frequência de cada uma.
#ele conta a frequência de cada valor único em uma coluna
#"Eu tenho uma caixa cheia de bolinhas de cores misturadas,
#me diga quantas de cada cor tem." (você não sabe as cores nem as quantidades de antemão)

plt.bar(mais_falhas.index,mais_falhas.values)
plt.xlabel("Empresa")
plt.ylabel("Número de falhas")
plt.title("Top 5 empresas com mais falhas")
plt.xticks(rotation=45)
plt.show()



```

Questão 5


Por meio do dataset paises.csv, trace um gráfico de dispersão
relacionando a renda per capita (GDP ($ per capita)) com a taxa de
alfabetização (Literacy (%)) dos países da América Latina. Além disso,
faça com que o tamanho de cada ponto represente a população do país;


```python
mask_regiao = ds['Region'] == "LATIN AMER. & CARIB"

dados_regiao = ds[mask_regiao]

plt.figure(figsize=(10, 6))



plt.scatter(dados_regiao['GDP ($ per capita)'], dados_regiao['Literacy (%)'],
            s=dados_regiao['Population']/dados_regiao['Population'].max()*1000,
            alpha=0.5)
plt.xlabel("Renda per capita")
plt.ylabel("Taxa de alfabetização")
plt.title("Relação entre renda per capita e taxa de alfabetização dos países da América Latina\n(tamanho do ponto = população)")
plt.show()
```

Questão 6

Com base nos dados do dataset space.csv, trace um gráfico em torta
ilustrando a porcentagem geral de foguetes com status “StatusActive” e
“StatusRetired”, considerando todas as empresas;


```python
mask_foguetes_active = data['Status Rocket'] == "StatusActive"
mask_foguetes_retired = data['Status Rocket'] == "StatusRetired"

dados_foguetes_active = data[mask_foguetes_active]
dados_foguetes_retired = data[mask_foguetes_retired]

n_active = dados_foguetes_active['Status Rocket'].count()
n_retired = dados_foguetes_retired['Status Rocket'].count()
#Ou :
#contagem = ds["Status Rocket"].value_counts()
#n_active = contagem["StatusActive"]
#n_retired = contagem["StatusRetired"]
#.count() : você já isolou um grupo específico e só quer saber o tamanho desse grupo
#"Eu já separei as bolinhas vermelhas em uma pilha à parte.
#Quantas bolinhas tem nessa pilha?" (você já sabe que é só vermelho, só quer o total)

porcentagem_active = (n_active / (n_active + n_retired)) * 100
porcentagem_retired = (n_retired / (n_active + n_retired)) * 100

fatias = plt.pie([porcentagem_active,porcentagem_retired],autopct="%1.1f%%")
plt.legend(labels=['StatusActive', 'StatusRetired'])
plt.title('Porcentagem de foguetes ativos e inativos')

plt.show()

```

Questão 7

Por meio do dataset paises.csv, trace em uma mesmo plano cartesiano
duas linhas com estilos customizados (marcador, cor e tipo de linhas
diferentes) comparando o “GDP ($ per capita)” e o número de “Phones
(per 1000)” dos países da Europa Ocidental (WESTERN EUROPE);


```python
mask_europa = ds['Region'] == "WESTERN EUROPE"

dados_europa = ds[mask_europa]

plt.figure(figsize=(16,6))

plt.plot(dados_europa['Country'],dados_europa['GDP ($ per capita)'],label="GDP ($ per capita)",
         marker="o", color="blue", linestyle="-")

plt.plot(dados_europa['Country'],dados_europa['Phones (per 1000)'],label="Phones (per 1000)",
         marker="s", color="red", linestyle="--")

plt.xlabel("Country")

plt.xticks(rotation=45, ha= "right",fontsize = 8)



plt.title("GDP per capita e Phones per 100 dos países da europa ocidental")
plt.legend()
plt.show()

print("Escala do telefone é muito maior, o que achata linha pro 0")
dados_europa['Phones (per 1000)'].describe()


```

Questão 8

Por meio do dataset space.csv, usando o conceito de subplot(), trace
dois gráficos de barras separados lado a lado: à esquerda, as 5 empresas
com mais missões de sucesso; à direita, as 5 empresas com mais missões
de falha.

```python
mask_failure = data['Status Mission'] == "Failure"
mask_sucess = data['Status Mission'] == "Success"

dados_failure = data[mask_failure]
dados_sucess = data[mask_sucess]

mais_sucessos = dados_sucess["Company Name"].value_counts().nlargest(5)

mais_falhas = dados_failure["Company Name"].value_counts().nlargest(5)

plt.figure(figsize=(16,6))

plt.subplot(1,2,1)
plt.bar(mais_sucessos.index,mais_sucessos.values)
plt.xlabel("Empresa")
plt.ylabel("Número de sucessos")
plt.title("Top 5 empresas com mais sucessos")
plt.xticks(rotation=45,ha="right")

plt.subplot(1,2,2)
plt.bar(mais_falhas.index,mais_falhas.values)
plt.xlabel("Empresa")
plt.ylabel("Número de falhas")
plt.title("Top 5 empresas com mais falhas")
plt.xticks(rotation=45,ha="right")

plt.show()


```
