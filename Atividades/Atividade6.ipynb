```python
import pandas as pd
import numpy as np

data = pd.read_csv("datasets/paises.csv",sep=";")

display(data)

print(f"Colunas do dataset : \n{data.columns}")
```

Questão 1

```python
data["Region"] = data["Region"].str.strip() #Retira espaços em branco do início e do fim da string
data_regiao = data.groupby("Region").get_group("OCEANIA")

#Ou
#mascara = data["Region"].str.contains("OCEANIA")
#data_regiao = data[mascara]

print(f"Quantidade de países na região da Oceania: {len(data_regiao)}")
print(f"Nomes dos países da região da Oceania: \n{data_regiao['Country']}")
```

Questão 2

```python
pais_mais_populoso = data.nlargest(1, "Population")
print(f"População do país mais populoso: \n{pais_mais_populoso['Population']}")
print(f"Região do país mais populoso: \n{pais_mais_populoso['Region']}")
```

Questão 3

```python
regioes = data.groupby("Region")

alfabetizacao_regiao = regioes["Literacy (%)"].mean()


alfabetizacao_regiao = alfabetizacao_regiao.round(2)

for regiao, alfabetizacao in alfabetizacao_regiao.items():
    print(f"Região: {regiao} - Taxa de alfabetização média: {alfabetizacao}%")

##Jeito sem usar for
alfabetizacao_regiao = data.groupby("Region")["Literacy (%)"].mean().round(2)

saida = "Regiao: " + alfabetizacao_regiao.index + " - Taxa de alfabetização média: " + alfabetizacao_regiao.astype(str) + "%"
print(saida.to_string(index=False))

```

Questão 5

```python
def deathrate_medidor(deathrate):
    if deathrate < 9:
        return "Balanced"
    else:
        return "Urgent"


taxa_mortalidade = data["Deathrate"]

data["Humanitarian Help"] = taxa_mortalidade.apply(deathrate_medidor)

display(data)
```

Questão 6

```python
data_regiao = data.groupby("Region")

print(f"Estatísticas descritivas da população por região: \n{data_regiao['Population'].describe()}")
print(f"Estatísticas descritivas da população por região(Apenas 5 primeiras linhas): \n{data_regiao['Population'].describe().head(5)}")
```

Questão 7

```python
def meta_reduzir_mortaludade_infatil(deathrate_kids):
    return deathrate_kids - deathrate_kids * 0.15

data["Infant mortality meta"] = data["Infant mortality (per 1000 births)"].apply(meta_reduzir_mortaludade_infatil)

comparacao = pd.concat([data["Infant mortality (per 1000 births)"], data["Infant mortality meta"]], axis=1)

print(f"Comparação entre a taxa de mortalidade infantil atual e a meta de redução: \n{comparacao}")
```
