```python
import numpy as np
import pandas as pd
```

```python
#Questão 1
lista_linguagem = ["Java","C","Python",]
lista_porcentagem = [16.25,16.04,9.85]

seriesAno1 = pd.Series(index = lista_linguagem,data = lista_porcentagem)
print(f"Series primeiro ano : \n{seriesAno1}")

lista_linguagem = ["C","Python","Java",]
lista_porcentagem = [16.21,12.12,11.68]

seriesAno2 = pd.Series(index = lista_linguagem,data = lista_porcentagem)
print(f"Series segundo ano : \n{seriesAno2}")

#Ou criava usando
#seriesAno1 = pd.Series({'Java': 16.25, 'C': 16.04, 'Python': 9.85})
#seriesAno2 = pd.Series({'C': 16.21, 'Python': 12.12, 'Java': 11.68})

```

```python
#Questão 2

porcentagem_total_primeiro_ano = seriesAno1.sum()
print(f"Porcentagem somada linguagens ano um : {porcentagem_total_primeiro_ano}%")

porcentagem_total_segundo_ano = seriesAno2.sum()
print(f"Porcentagem somada linguagens ano dois : {porcentagem_total_segundo_ano}%")
```

```python
#Questão 3

print(f"Variação porcentagem entre os anos de cada linguagem : \n{seriesAno2.subtract(seriesAno1)}")
```

```python
#Questão 4

crescimento = seriesAno2.subtract(seriesAno1)
condicao = crescimento > 0

print(f"Linguagens que tiverem crescimento positivo : \n{crescimento[condicao]}")
#Ou usa crescimento[crescimento>0]
```

```python
#Questão 5

projecao = seriesAno2 + 2*crescimento

print(projecao)

mais_popular = projecao.nlargest(1)

print("Linguagem mais popular se taxas mantiverem constantes após 2 anos : ")
print(mais_popular)
```

```python
#Exercício 6

coluna = ["W","X","Y","Z"]
linhas = ["A","B","C","D","E"]

np.random.seed(10)
valores = np.random.randint(1,50,[5,4])
df = pd.DataFrame(columns=coluna,index=linhas,data=valores)

print("Dataframe gerado : ")
display(df)

condicao = df["X"] < 30

elementos_validos =  df["X"][condicao]

print(f"Média elementos menores que 30 : \n{elementos_validos.mean()}")

```

```python
#Questão 7

elementos_D = df.loc["D"]
print(f"Média elementos linha D : {elementos_D.mean():.2f}")

elementos_E = df.iloc[4,:]
#elementos_E = df.iloc[4] tb dá
soma_linhaE = elementos_E.sum()
print(f"Soma elementos linha E : {soma_linhaE}")
```
