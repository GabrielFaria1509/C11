Questão 1

```python
coluna_idade = dataset[:,1].astype(int)

media_idades = np.mean(coluna_idade)

print(f"Média das idades : {media_idades:.2f}")

```

Questão 2

```python
coluna_genero = dataset[:,2]

coluna_gastos = dataset[:,5].astype(float)

condicao = np.char.find(coluna_genero,"Male") != -1

coluna_homens = coluna_genero[condicao]

coluna_gastos_homens = coluna_gastos[condicao]

print(coluna_gastos_homens)

media_gastos = np.mean(coluna_gastos)

print(f"Média dos gastos : {media_gastos:.2f}")

cond = (coluna_gastos_homens > media_gastos) != -1

homens_validados = coluna_gastos_homens[cond]

print(f"Quantidade de homens que gastaranm mais que a média : {len(homens_validados)}")


```

Questão 3

```python
coluna_frequencia_compra = dataset[:,15].astype(int)

print(coluna_frequencia_compra)

total_vendas = np.sum(coluna_frequencia_compra)

produto_menos_vendido = min(coluna_frequencia_compra)

porcentagem = (produto_menos_vendido/total_vendas)*100

print(f"Porcentagem do produto com menos vendas : {porcentagem:.2f}%")










```

Questão 4

```python
coluna_desconto = dataset[:,13]

condicao = np.char.find(coluna_desconto,"Yes") != -1

coluna_desconto_usado = coluna_desconto[condicao]

quantidade_desconto_usado = len(coluna_desconto_usado)

print(f"Quantidade de descontos usados : {quantidade_desconto_usado}")

quantidade_total_coluna_desconto = len(coluna_desconto)

porcentagem_desconto_usado = (quantidade_desconto_usado/quantidade_total_coluna_desconto)

print(f"Porcentagem de descontos utlizados em relaçao ao total : {porcentagem_desconto_usado*100:.2f}%")








```
