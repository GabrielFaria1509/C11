```python
coluna_regioes = dataset[:,1]

coluna_paises = dataset[:,0]

coluna_renda_per_capita = dataset[:,8].astype(float)  ##Puxo as colunas que vo precisar

condicional = np.char.find(coluna_regioes,"LATIN AMER. & CARIB") != -1 #Condição para delimitar região

coluna_regiao_desejada = coluna_regioes[condicional]  #Limito as colunas pra dentro da região

coluna_paises_regiao_desejada = coluna_paises[condicional]

rendas_regiao_desejada = coluna_renda_per_capita[condicional]

index_pais_maior_renda = np.argmax(rendas_regiao_desejada) #Descubro index do país

pais_maior_renda = coluna_paises_regiao_desejada[index_pais_maior_renda]  



print(f"A maior renda per capita da América do SUL e Caribe pertence ao país : \n{pais_maior_renda}")

```
