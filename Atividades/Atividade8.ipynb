```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

Questão 1

Utilizando o dataset Iris do Seaborn, para os dados apenas da espécie ‘setosa’, avalie a correlação entre suas variáveis numéricas, apresentando
os valores das correlações e permitindo identificar quais variáveis estão
mais correlacionadas

```python
ds = sns.load_dataset('iris')
print(f"Colunas do dataset iris : \n{ds.columns}")

corr = ds[["sepal_length", "sepal_width", "petal_length", "petal_width"]].corr()
sns.heatmap(corr,annot = True,fmt = ".2f")
plt.show()
```

Questão 3

Utilizando novamente o dataset titanic, analise a distribuição das idades
dos passageiros em função da classe, segmentando os dados por sexo, de
modo a construir uma visualização que permita comparar a tendência
central, a dispersão e possíveis valores extremos entre os grupos

```python
ds_titanic = sns.load_dataset('titanic')

ds_titanic = ds_titanic.dropna(subset="age")

sns.boxplot(
    data = ds_titanic,
    x = "class",
    y = "age",
    hue = "sex",
    palette="Set2",
    showmeans=True, #Destaca mediana
    meanprops={"marker": "D", "markerfacecolor": "white",
               "markeredgecolor": "black", "markersize": 6},
)

plt.title("Distribuição das idades dos passageiros em função da classe")
plt.show()
```

Questão 4

Utilizando do dataset mpg, analise a relação entre a potência dos
veículos (horsepower) e o consumo de combustível (mpg), construindo uma
visualização que permita identificar a tendência entre essas variáveis e
interpretar seu comportamento

```python
ds_mpg = sns.load_dataset("mpg").dropna(subset=["horsepower", "mpg"])

sns.regplot(data = ds_mpg, x = "horsepower", y = "mpg")
plt.xlabel("Potência (horsepower)")
plt.ylabel("Consumo (milhas por galão, mpg)")
plt.title("Relação entre a potência dos veículos e o consumo de combustível")
plt.show()
```

Questão 5

5. Utilizando o dataset paises.csv, calcule a correlação entre
“GDP ($ per
capita)”, “Literacy (%)”, “Infant mortality (per 1000 births)” e
“Phones (per
1000)”, apresentando o resultado em um Heatmap com os valores numéricos
exibidos;


```python
dataset = pd.read_csv("paises.csv", sep = ";")

print(f"Colunas do dataset : \n{dataset.columns}")

cols = ["GDP ($ per capita)", "Literacy (%)",
        "Infant mortality (per 1000 births)", "Phones (per 1000)"]

corr = dataset[cols].corr()
sns.heatmap(corr,annot = True,fmt = ".2f")
plt.title("Correlação entre indicadores socioeconômicos")

plt.show()
```

Questão 6

Com o dataset paises.csv, trace um Boxplot comparando a distribuição do “GDP ($ per capita)” entre os países da América Latina e da Europa
Ocidental, permitindo assim identificar mediana, dispersão e outliers de
cada grupo;

```python
dataset['Region'] = dataset["Region"].str.strip()  ##Tirando espaço em branco

paises_america_latina = dataset[dataset["Region"] == "LATIN AMER. & CARIB"]
paises_europa_ocidental = dataset[dataset["Region"] == "WESTERN EUROPE"]
#Ou usar : regioes = ["LATIN AMER. & CARIB", "WESTERN EUROPE"]
#df_plot = dataset[dataset["Region"].isin(regioes)]

paises_regioes = pd.concat([paises_america_latina, paises_europa_ocidental])


sns.boxplot(data = paises_regioes, x = "Region", y = "GDP ($ per capita)",palette = "Set2")
#sns.boxplot(data = df_plot, x = "Region", y = "GDP ($ per capita)",palette = "Set2")
plt.xlabel("Região")
plt.ylabel("Renda per capita (US$)")
plt.title("Distribuição do GDP ($ per capita) entre os países da América Latina e da Europa Ocidental")
plt.show()

```

Questão 8

Por meio do dataset space.csv, após filtrar apenas as missões com custo
conhecido (Cost > 0), trace um Boxplot comparando a distribuição do custo
das missões
(Cost) entre foguetes com status
“StatusActive” e
“StatusRetired”, permitindo identificar mediana, dispersão e outliers de
cada grupo.

```python
data_space = pd.read_csv("space.csv", sep = ";")
print(f"Colunas do dataset : \n{data_space.columns}")
#Retirando espaço em branco do título da coluna de custo e outros

data_space = data_space.rename(columns={" Cost": "Cost"})
#Ou data_space.columns = data_space.columns.str.strip()

#Passo pra número os valores, coerce é quem não conseguir vira NaN
data_space['Cost'] = pd.to_numeric(data_space['Cost'], errors='coerce')
#Retiro valores vazios
data_space = data_space.dropna(subset=["Cost"])

mask_custos = data_space['Cost'] > 0
data_space = data_space[mask_custos]

#versão mais recente do seaborn pede usar hue e palette em conjunto
sns.boxplot(data = data_space, x = "Status Rocket", y = "Cost", palette = "Set2",hue = "Status Rocket")
plt.xlabel("Status")
plt.ylabel("Custo")
plt.title("Distribuição do custo das missões entre foguetes com status 'StatusActive' e 'StatusRetired'")
plt.show()

```
