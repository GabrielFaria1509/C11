```python
# Exercício 1

coluna_status = dataset[:,7]
missoes_sucesso = np.sum(coluna_status=="Success")

missoes_sucesso_porcentagem = (missoes_sucesso/len(coluna_status))*100
print(f"Missões com sucesso : \n{missoes_sucesso_porcentagem:.2f}")

```

```python
#Exercício 2

coluna_gastos = dataset[:,6].astype(float) 

mascara = coluna_gastos != 0

gastos = coluna_gastos[mascara]

media_gastos = np.mean(gastos)

print(f"Média dos gastos : \n{media_gastos}")

```

```python
#Exercício 3

missoes_locais = dataset[:,2]

cond = np.char.find(missoes_locais,"USA") != -1

print(cond)

missoes_EUA = missoes_locais[cond]

print(missoes_EUA)

print(f"Quantidade de missões nos EUA : \n{len(missoes_EUA)}")
```
