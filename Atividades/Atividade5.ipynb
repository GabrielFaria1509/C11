```python
# Exercício 1

coluna_status = dataset[:,7]
missoes_sucesso = np.sum(coluna_status=="Success")

missoes_sucesso_porcentagem = (missoes_sucesso/len(coluna_status))*100
print(f"Missões com sucesso : \n{missoes_sucesso_porcentagem:.2f}%")

```

```python
#Exercício 2

coluna_gastos = dataset[:,6].astype(float) 

mascara = coluna_gastos != 0

gastos = coluna_gastos[mascara]

media_gastos = np.mean(gastos)

print(f"Média dos gastos : \n{media_gastos}")

```

```python
#Exercício 3

missoes_locais = dataset[:,2]

cond = np.char.find(missoes_locais,"USA") != -1

print(cond)

missoes_EUA = missoes_locais[cond]

print(missoes_EUA)

print(f"Quantidade de missões nos EUA : \n{len(missoes_EUA)}")
```

```python
# Exercício 4
missoes_empresas =dataset[:,1]

condicional = np.char.find(missoes_empresas,"SpaceX") != -1

missoes_custo = dataset[:,6].astype(float) 

detalhesmissoes = dataset[:,4]

missoes_SpaceX = missoes_empresas[condicional]

custosSpaceX = missoes_custo[condicional]

maior_custo_spaceX = np.max(custosSpaceX)

detalhesSpaceX = detalhesmissoes[condicional][np.argmax(maior_custo_spaceX)]

print(f"O missão mais custosa para a SpaceX teve custo de \n {maior_custo_spaceX}")
print(f"Detalhes missão : \n{detalhesSpaceX}")
```

```python
#Exercício 5

missoes_empresas =dataset[:,1]

quantidade_empresas,contagem = np.unique(missoes_empresas,return_counts="True")

for empresa,qtd in zip(quantidade_empresas,contagem):
    print(f"A empresa {empresa} realziou {qtd} missões")

#\ou 
 

#for i in range(0, len(quantidade_empresas)):
    #print(f"Empresa {quantidade_empresas[i]} realizou {contagem[i]} missões")

```

```python
# Exercício 6

coluna_statusrocket = dataset[:,5]

missoes_retired = np.sum(coluna_statusrocket=="StatusRetired")

porcentagem_retired = (missoes_retired/len(coluna_statusrocket))*100

print(f"Porcentagem missões com status do foguete como retired : \n{porcentagem_retired:.2f}%")




```

```python
#Exercício 7
missoes_locais = dataset[:,2]

condition = np.char.find(missoes_locais,"Russia") != -1

missoes_Russia = missoes_locais[condition]

print(f"Quantidade de missões na Russia : \n{len(missoes_Russia)}")
```

```python
#Exercício 8

coluna_gastos = dataset[:,6].astype(float)

mascara = coluna_gastos != 0

gastos = coluna_gastos[mascara]

maior_gasto = np.max(gastos)

index_maior_gasto = np.argmax(coluna_gastos)

print(f"Maior gasto isolado : \n{maior_gasto:.2f}")

coluna_empresas = dataset[:,1]

empresa_maior_gasto = coluna_empresas[index_maior_gasto]

print(f"Empresa com maior gasto : \n{empresa_maior_gasto}")
```
